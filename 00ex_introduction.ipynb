1. The MickeyMouse problem

a) Write a program that prints the numbers from 1 to 100. But for multiples of 3 print `Mickey` instead of the corresponding number and for the multiples of 5 print `Mouse`. For numbers which are multiples of both three and five print `MickeyMouse`

b) Put the result in a tuple and substitute `Mickey` with `Donald` and `Mouse` with `Duck`

```python
divisors = { 3 : "Mickey", 5: "Mouse" }
keys = divisors.keys()

i = 1
while i <= 100: 
    word = ""
    for j in keys:
        if not i%j: word += divisors[j]
    if word == "": print(i)
    else: print(word)
    i+=1

# [ ] of course better/other ways to do this are possible, explore them
```

3\. Computing the distance

Write a function that calculates and returns the euclidean distance between two points *u* and *v*, where *u* and *v* are both 2-tuples *(x,y)*. For example, if *u=(3,0)* and *v=(0,4)*, the function should return 5

```python
import math

def eudistance(x1,x2):
    dis = math.sqrt((x1[0]-x2[0])**2  + (x1[1]-x2[1])**2)
    return dis

x1 = (3,0)
x2 = (0,4)

distance = eudistance(x1, x2)
print(distance)
```

4\. Counting letters

Write a program to calculate the number of times each character occurs in a given string *s*. Ignore differneces in capitalization

```python
s="Write a program that prints the numbers from 1 to 100. \
But for multiples of three print Mickey instead of the number and for the multiples of five print Mouse. \
For numbers which are multiples of both three and five print MickeyMouse"

def countchar(str, char):
    if char.lower() == char:
        lowchar = char
        upchar = char.upper()
    else:
        lowchar = char.lower()
        upchar = char
    return str.count(lowchar) + str.count(upchar)

print("Count for h: ",  countchar(s,"h"))
print("Count for H: ",  countchar(s,"H"))

print("Count for s: ",  countchar(s,"s"))
print("Count for S: ",  countchar(s,"S"))

print("Count for l: ",  countchar(s,"l"))
print("Count for L: ",  countchar(s,"L"))

# [ ] try again without using the count method with direct string manipulation/operators
```

5\. Isolating the unique

Write a function that determines and count the unique numbers in the list *l*

```python
l = [36, 45, 58, 3, 74, 96, 64, 45, 31, 10, 24, 19, 33, 86, 99, 18, 63, 70, 85,
 85, 63, 47, 56, 42, 70, 84, 88, 55, 20, 54, 8, 56, 51, 79, 81, 57, 37, 91,
 1, 84, 84, 36, 66, 9, 89, 50, 42, 91, 50, 95, 90, 98, 39, 16, 82, 31, 92, 41,
 45, 30, 66, 70, 34, 85, 94, 5, 3, 36, 72, 91, 84, 34, 87, 75, 53, 51, 20, 89, 51, 20]
```

```python
# using sets: the function returns the new count and updates the list (exercise did not necessarily say to keep a list
# so one could just keep the set 
def uniquelist(alist):
    unique = set()
    for n in alist:
        unique.add(n)
    alist = list(unique)
    return len(alist)

print("l before:", l)
print("Elements before:", len(l))
print("Elements after", uniquelist(l))
print("l after:", l)

# [ ] try another solution that does not involve the sets
```

6\. Combination of functions

Write two functions - one that returns the square of a number, and one that returns the cube. Now write a third function that returns the number raised to the 6th power using the two previous functions.

```python
def square(x):
    return x*x

def cube(x):
    return x**3

for i in range(5):
    print("Square of", i, ":", square(i))
    print("Cube of", i, ":", cube(i))
    print("Sixth power of", i,":", square(cube(i)))
```

7\. Cubes

Create a list of the cubes of x for x in *[0, 10]* using:

```python
# first using a for loop
cubelist = []

for x in range(11):
    cubelist.append(cube(x))

print(cubelist)

# now with list comprehension
cubelist = []
cubelist = [cube(x) for x in range(11)]
print(cubelist)
```

8\. Nested list comprehension

A Pythagorean triple is an integer solution to the Pythagorean theorem $a^2+b^2=c^2$. The first Pythagorean triple is (3,4,5). Find and put in a tuple all unique Pythagorean triples for the positive integers a, b and c less than 100.

```python
pytouples = []

maxnum = 100
for a in range(maxnum):
    for b in range(maxnum):
        for c in range(maxnum):

            if a*a + b*b == c*c:
                pytouples.append((a,b,c))

print(pytouples)
```

9\. Normalization

Write a function that takes a tuple of numbers and returns it with the entries normalized to one

```python
def normalizetouple(vec): 
    norm = 0.0
    for x in vec:
        norm += x*x
    normalized = []
    for x in vec: 
        normalized.append(x/math.sqrt(norm))
    return tuple(normalized)

t1 = (1,2,3)
t2 = (1,2,2,2)
t3 = (1/math.sqrt(2), 1/math.sqrt(2))

print(normalizetouple(t1))
print(normalizetouple(t2))
print(normalizetouple(t3))

# [ ] is there a way to do it without having to conver ot a list to a tuple? try
```
