
````python
def hellodecorator(func,x):
    def wrapper():
        print("Hello!")
        func(x)
    return wrapper

@hellodecorator
def square(x):
    return x*x

square(10)
````

4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
# factorial with recursion
def recfactorial(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return n*recfactorial(n-1)


# factorial without recursion
def norecfactorial(n): 
    if n == 0:
        return 0
    else:
        result = n
        i = n-1
    while i > 0:
        result = result*i
        i -= 1
    return result

# let's compute some factorials to see if it is correct
nums = [0, 1, 2, 5]
for i in nums: print("recursive factorial result: ", recfactorial(i), "non recursive factorial result: ", norecfactorial(i))
    
````
