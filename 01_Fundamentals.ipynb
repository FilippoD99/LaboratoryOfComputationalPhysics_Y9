# Copies and Views

#### Copies:
* A **copy** creates a new object that is a duplicate of the original one.
* Changes made to the copy do not affect the original object, and vice versa.
* There are two types of copies:
  * Shallow Copy: Only the top-level object is copied. If the original object contains references to other objects (e.g., a list of lists), the references are copied, not the objects themselves.
  * Deep Copy: A complete copy is made, including all nested objects. Changes to any part of the deep copy won’t affect the original object.

```python
# further info on variables in python: we saw that they are all pyobject
# and can be immutable and mutable
# no explicit pointers in python
# but there is something that resembles that
# shallow copy: change does not affect original. subtletly: with complex object that my have a nested 
# structure (like a dict of a dict ...). in this case copy can be shallow or nested: shallow does not 
# resolve the nested structure: just the top level obj is copied (memory managment feature)
# but this is quite dangerous 
# copy library. nb: when we'll use pd or complex data structure you'll have to check what a method does
# in terms of copies and views: need to check yourself

import copy
original = [1, 2, [3, 4]]

shallow_copy = copy.copy(original)
shallow_copy[0] = 100  # This won't affect the original
shallow_copy[2][0] = 300  # This will affect the original's nested list

print("Original:", original)  
print("Shallow Copy:", shallow_copy) 
```

#### Views:

not a pointer: just an object with different name but pointing to the same memory location. different naming for an object you already created (view).

* A view provides a reference to the original object without creating a new object.
* Changes made through the view directly affect the original object because the view is simply another way of accessing the original data.
* We will see in the next classes that NumPy arrays and Pandas DataFrames often deal with views when you slice or manipulate them.


```python
# slicing this numpy array gets you a view, not a copy: so changing it changes the object
# operating on the original object: choice sometimes one makes in implementing a function/algorithm
# rational behind it is that numpy arrays are meant to frame complex large dataset/datastructures
# very large data object: don't want to run the risk of having issues with memory management: we don't
# want a new memory allocation every time we make operations on it: so you provide a view
# choice suitable for numpy. other object might have different goals and one might choose copies
# typically with large data structures one wants to provide views.
# need to be aware of the difference
import numpy as np
arr = np.array([1, 2, 3, 4])
view = arr[1:3]  # This creates a view, not a copy

view[0] = 100  # Modifies the original array

print("Original array:", arr)  
print("View:", view)  
```

## Functions arguments

In Python, whether arguments passed to a function are treated as views or copies depends on the data type of the argument and how it is handled inside the function. Here's how Python handles various types of arguments:

**Immutable Types** (e.g., integers, strings, tuples):
* For immutable types like `int`, `str`, and `tuple`, when you pass them as arguments to a function, Python creates a copy of the reference to the object, not the object itself.
* Since these objects cannot be modified, any attempt to change them inside the function results in the creation of a new object, leaving the original object unchanged.

```python
# what happens when i pass a variable to a function? depends on the datatype, it's up to the scope
# of the function to define how one deals with it
# if pass an immutable object python creates a copy: modify a copy (basically pass by value)
# within the function the var is a copy, if modify you modify the copy
def modify(x):
    x += 1
    print("Inside function:", x)

a = 10
modify(a)
print("Outside function:", a)
# but what if i return?  try [ ]
```

**Mutable Types** (e.g., lists, dictionaries, sets):
* For mutable types like `list`, `dict`, or `set`, Python passes a reference to the original object. This means that any modification made to the object inside the function will affect the original object outside the function, as they both reference the same object (this behaves like a view).

```python
# with mutable obj the original object gets modified
def modify_list(lst):
    lst.append(4)
    print("Inside function:", lst)

my_list = [1, 2, 3]
modify_list(my_list)
print("Outside function:", my_list)
```

## Higher order functions

A function that uses another function as an input argument or returns a function is known as a higher-order function (HOF). The most familiar examples are `map` and `filter`.

### map

The map function applies a function to each member of a collection

```python
# map and filter: very useful. for example apply square to every element in a range
# nb: i added the function (should it have been there in the first place?
def square(x):
    return x*x
    
x = list(map(square, range(5))) 
print (x)

# Note the difference w.r.t python 2. In python 3 map retuns an iterator so you can do stuff like:
for i in map(square,range(5)): print(i)

# or
[i for i in map(square,range(6))]
```

### filter

The filter function applies a predicate to each member of a collection, retaining only those members where the predicate is True

```python
# need a boolean to decide if retain element
def is_even(x):
    return x%2 == 0

print (list(filter(is_even, range(5))))
```

Combinations in sequence of HOF are obviously possible

```python
# square of even numbers only: map square after a filter is applied
list(map(square, filter(is_even, range(5))))
```

## Recursive functions 

```python
def fib1(n):
    """Fib with recursion."""

    # base case
    if n==0 or n==1:
        return 1
    # recursive case
    else:
        return fib1(n-1) + fib1(n-2)

    
print ([fib1(i) for i in range(10)])
```

```python
# important thing: there are several ways to implement an operation/algorithm, as long as the algorithm
# is represented correctly (right output) there is no right or wrong. in general it is right
# even mathematically the same concept holds of course. 
# BUT: given the fact that computer do very complex computations, different implementation of the same 
# logic has different execution time, one is more appropriate than others
# time taking is not only parameter, also how much memory used: time mangement and memory management.

# In Python, a more efficient version that does not use recursion is

def fib2(n):
    """Fib without recursion."""
    a, b = 0, 1
    for i in range(1, n+1):
        a, b = b, a+b
    return b

print ([fib2(i) for i in range(10)])
```

```python
# check indeed the timing:

%timeit fib1(20)
%timeit fib2(20)
```

## Decorators

Decorators are a type of HOF that take a function and return a wrapped function that provides additional useful properties.

Examples:

  - logging
  - profiling
  - Just-In-Time (JIT) compilation

```python
# syntax a lot of people like in python. enhance a given function with additional properties
# taking a function as input and make wrapper and return i
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper

def say_whee():
    print("Whee!")

say_whee = my_decorator(say_whee)
```

```python
say_whee()
```

Python allows you to use decorators in a simpler way with the @ symbol, sometimes called the “pie” syntax

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper

@my_decorator
def say_whee():
    print("Whee!")
```

```python
say_whee()
```

#### JIT

A JIT (Just-In-Time) compiler refers to a technique used to improve the performance of code execution by compiling code into machine code at runtime, rather than interpreting it line by line. This helps speed up program execution by reducing the overhead of repeatedly interpreting code.

While the standard Python interpreter, CPython, does not include a JIT compiler, there are alternative Python implementations that provide JIT compilation, such as PyPy or [numba](https://numba.pydata.org).

JIT combines both interpretation and compilation. It interprets code initially, and as it identifies sections of code that are executed repeatedly, it compiles them to machine code, optimizing performance dynamically.



```python
# python is typically inefficient as every interpreted lang.
# typically faster with jit. speed up python by providing compilation step inline
# rather sophisticated topic, just know it exist
# here decorating the sum with jit: compile in machine code
from numba import jit

@jit(nopython=True)
def fast_sum(arr):
    total = 0  # check this 
    for i in arr:
        total += i
    return total
```

```python
#fast_sum(list(range(100)))
#fast_sum([1,2,3,4,5])
import numpy as np

arr = np.arange(1e6, dtype=np.float64)
print(fast_sum(arr))

# is it really fast? Try to compare with my_add above
```
